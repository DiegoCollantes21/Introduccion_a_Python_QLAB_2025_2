## Sección 4: Estadística y Síntesis

**13.** Muestra un resumen estadístico completo de las columnas numéricas del DataFrame.

```python
# Columnas numéricas
df_num = df_desarrollo.select_dtypes(include='number')

# Base de percentiles
base = df_num.describe(percentiles=[.05, .25, .5, .75, .95]).T

# Desviación absoluta mediana y Coeficiente de variación
mad = (df_num.sub(df_num.median())).abs().mean()
cv = df_num.std(ddof=1) / df_num.mean()

# Asimetría y curtosis
skew = df_num.skew(numeric_only=True)
kurt = df_num.kurtosis(numeric_only=True)

# IQR
q1 = df_num.quantile(0.25)
q3 = df_num.quantile(0.75)
iqr = q3 - q1

# Combinar todo en un solo DataFrame
resumen_completo = (
    base
    .assign(
        MAD=mad,
        CV=cv,
        asimetria=skew,
        curtosis=kurt,
        IQR=iqr
    )
    # Orden sugerido de columnas
    .reindex(columns=[
        'count','mean','std','min','5%','25%','50%','75%','95%','max',
        'IQR','MAD','CV','asimetria','curtosis'
    ])
    .round(4)  # redondeo para legibilidad
)

print(resumen_completo)
```

**14.1.** Calcula la matriz de correlación para las columnas numéricas del DataFrame.

**14.2.** **Interpretación:** Identifica el par de variables con la correlación positiva más fuerte y el par con la correlación negativa más fuerte (excluyendo la correlación de una variable consigo misma). ¿Tienen sentido estos resultados en el mundo real?

```python
# Matriz de correlación (Pearson)
corr = df_num.corr(method='pearson', numeric_only=True)
print('Matriz de correlación:\n', corr, '\n')

mask = np.triu(np.ones(corr.shape), k=1).astype(bool)  # parte superior sin diagonal
corr_upper = corr.where(mask)

# Par positivo más fuerte
max_pos_val = corr_upper.stack().idxmax()
max_pos_corr = corr_upper.stack().max()

# Par negativo más fuerte
min_neg_val = corr_upper.stack().idxmin()
min_neg_corr = corr_upper.stack().min()

print(f'Máxima correlación positiva: {max_pos_val} = {max_pos_corr:.4f}')
print(f'Máxima correlación negativa: {min_neg_val} = {min_neg_corr:.4f}')
```

Tiene sentido que el PBI per cápita y la esperanza de vida se muevan juntos: cuando la gente gana más, suele tener mejor acceso a salud, buena alimentación y servicios, y eso se refleja en vidas más largas.
También es lógico que el índice de Gini se lleve mal con la esperanza de vida y con la alfabetización. En países más desiguales, los promedios de salud y educación tienden a ser peores.
En cambio, el tamaño de la población no dice mucho por sí solo sobre la riqueza por persona ni sobre la salud. En tu muestra, países como Japón, España o Canadá empujan hacia arriba la esperanza de vida y la alfabetización, mientras que casos como Brasil o México moderan la relación con el ingreso por habitante.
